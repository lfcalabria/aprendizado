```python
import pandas as pd
import warnings
warnings.filterwarnings("ignore")

from pycaret.datasets import get_data
from pycaret.classification import *
from ydata_profiling import ProfileReport

# 1. Carregar o dataset clássico Titanic
data = get_data('titanic')
print("Dataset carregado. Shape:", data.shape)
print(data.head())

# 2. EDA com ydata-profiling (sem dark_mode, minimal=False)
print("Gerando relatório EDA... Isso pode demorar alguns minutos.")
profile = ProfileReport(data, minimal=False)
profile.to_file("eda_titanic.html")
print("Relatório EDA salvo em: eda_titanic.html")

# 3. Pré-processamento e Setup do PyCaret
# PyCaret cuida de: imputação de missing values, normalização, encoding categórico,
# transformação, etc.
print("\nConfigurando ambiente PyCaret...")
setup(
    data=data,
    target='Survived',
    session_id=123,
    normalize=True,
    transformation=True,
    fold=10,
    log_experiment=False,
    verbose=True,
    use_gpu=False
)
'''
Classification ModelsThe classification module supports over 18 algorithms. Common estimators and their IDs include:L
inear: 
Logistic Regression (lr), Ridge Classifier (ridge), Linear Discriminant Analysis (lda).

Tree-based: 
Decision Tree (dt), Random Forest (rf), Extra Trees (et).

Boosting: 
AdaBoost (ada), Gradient Boosting (gbc), Extreme Gradient Boosting (xgboost), Light Gradient Boosting Machine (lightgbm), CatBoost (catboost).

Others: 
Naive Bayes (nb), K Neighbors (knn), Support Vector Machine (svm), Multi-layer Perceptron (mlp).
'''

# 4. Definir os 12 modelos suportados nativamente (mlp não é nativo no PyCaret 3.x)
models = [
    'lr', 'dt', 'rf', 'gbc', 'svm', 'knn', 'nb',
    'et', 'ada', 'catboost', 'xgboost', 'lightgbm'
]

# 5. AutoML: Comparação de métricas com Validação Cruzada 10-fold
print("\nComparando todos os modelos com CV 10-fold...")
top3 = compare_models(
    include=models,
    n_select=3,
    sort='AUC',  # Ordena por AUC (boa para classificação desbalanceada)
    fold=10
)
print("Top 3 modelos selecionados com sucesso.")

# 6. Voting Classifier com os top 3 modelos (soft voting)
print("Criando Voting Classifier (top 3)...")
voting_clf = blend_models(
    estimator_list=top3,
    method='soft',
    fold=10
)

# 7. Melhor modelo individual
best_model = top3[0]

# 8. Visualizações: Matriz de Confusão e ROC/AUC
print("Gerando plots...")
plot_model(best_model, plot='confusion_matrix')
plot_model(best_model, plot='auc')
plot_model(voting_clf, plot='confusion_matrix')
plot_model(voting_clf, plot='auc')

# 9. Finalizar modelos (treinamento no dataset completo)
print("Finalizando modelos...")
final_best = finalize_model(best_model)
final_voting = finalize_model(voting_clf)

# 10. Salvar modelos em arquivos pickle
save_model(final_best, 'melhor_modelo_titanic')
save_model(final_voting, 'voting_top3_titanic')
print("Modelos salvos:")
print("- melhor_modelo_titanic.pkl")
print("- voting_top3_titanic.pkl")

# 11. Exemplo de predições no conjunto de teste
print("\nExemplo de predições:")
pred_holdout = predict_model(final_best)
print(pred_holdout.head())

print("\nScript concluído com sucesso!")
```
