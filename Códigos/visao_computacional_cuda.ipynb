# 🧩 1. Detecção de Faces com OpenCV

Mesma detecção de rostos usando Haar Cascade. OpenCV continua usando CPU.


```python

!pip install opencv-python opencv-python-headless matplotlib -q

import cv2
import matplotlib.pyplot as plt

face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

def detect_faces(image_path):
    img = cv2.imread(image_path)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    faces = face_cascade.detectMultiScale(gray, 1.3, 5)
    for (x, y, w, h) in faces:
        cv2.rectangle(img, (x, y), (x + w, y + h), (255, 0, 0), 2)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    plt.imshow(img_rgb)
    plt.axis('off')
    plt.title(f'{len(faces)} rosto(s) detectado(s)')
    plt.show()

url = "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/lena.jpg"
!wget -q {url} -O lena.jpg

detect_faces("lena.jpg")
```


# 🧩 2. Reconhecimento de Objetos com YOLOv8 (GPU/CUDA)

O modelo vai usar GPU automaticamente se estiver disponível.


```python

!pip install ultralytics torch -q
import torch
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt

# Verifica GPU
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Dispositivo usado:", device)

# Carregar modelo YOLOv8 e mover para GPU se disponível
model = YOLO("yolov8n.pt")
model.to(device)

# Imagem de teste
img_url = "https://ultralytics.com/images/bus.jpg"
!wget -q {img_url} -O bus.jpg

results = model("bus.jpg", show=False)
res = results[0]
img = res.plot()
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.title("Reconhecimento de Objetos com YOLOv8 (GPU/CUDA)")
plt.show()

print("Objetos detectados:", res.names)
```


# 🧩 3. Leitura de Texto (OCR) com EasyOCR (GPU/CUDA)

EasyOCR também vai usar GPU automaticamente se disponível.


```python

!pip install easyocr torch -q
import easyocr

# Cria leitor usando GPU se disponível
gpu_available = torch.cuda.is_available()
reader = easyocr.Reader(['pt', 'en'], gpu=gpu_available)

print("GPU disponível para OCR?", gpu_available)

img_url = "https://i.imgur.com/3pO0n9I.png"
!wget -q {img_url} -O texto.png

result = reader.readtext("texto.png")

import cv2
import matplotlib.pyplot as plt

img = cv2.imread("texto.png")
for (bbox, text, conf) in result:
    (top_left, top_right, bottom_right, bottom_left) = bbox
    cv2.rectangle(img, tuple(map(int, top_left)), tuple(map(int, bottom_right)), (0, 255, 0), 2)
    cv2.putText(img, text, (int(top_left[0]), int(top_left[1]) - 10),
                cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 0, 0), 2)

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.title("Leitura de Texto com EasyOCR (GPU/CUDA)")
plt.show()

for (bbox, text, conf) in result:
    print(f"Texto detectado: {text} (confiança: {conf:.2f})")
```
